# Part 1 — Data Generation

**Week 8: E-Commerce Order Analytics System**

Generates the four CSV files required for the assignment:

1. `orders.csv`
2. `order_items.csv`
3. `products.csv`
4. `customers.csv`

The data contains intentional quality issues required by the assignment, including missing customer IDs, negative quantities, incorrect date formats, inconsistent product names, and invalid emails.

```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Configuration
NUM_ORDERS = 1000
NUM_ORDER_ITEMS = 2500
NUM_PRODUCTS = 100
NUM_CUSTOMERS = 500

random.seed(42)

# Create the data directory if it does not exist
import os
os.makedirs("data", exist_ok=True)
```

## 1. Define Sample Data

```python
categories = {
    "Electronics": ["Mobile", "Laptop", "Tablet", "Headphones", "Camera"],
    "Clothing": ["Shirts", "Jeans", "Shoes", "Jackets", "T-Shirts"],
    "Home": ["Furniture", "Kitchen", "Decor", "Bedding", "Appliances"],
    "Books": ["Fiction", "Education", "Technology", "Comics", "Biography"]
}

customer_types = ["REGULAR", "PREMIUM", "VIP"]

statuses = [
    "PLACED",
    "SHIPPED",
    "DELIVERED",
    "CANCELLED",
    "RETURNED"
]

regions = ["NORTH", "SOUTH", "EAST", "WEST"]
```

## 2. Generate Customers

```python
customers = []

start_registration = datetime(2024, 1, 1)

first_names = [
    "Rahul", "Amit", "Priya", "Sneha", "Rohan",
    "Neha", "Akash", "Pooja", "Aditya", "Anjali"
]

last_names = [
    "Sharma", "Patil", "Joshi", "Kulkarni",
    "Deshmukh", "Pawar", "More", "Jadhav"
]

for i in range(1, NUM_CUSTOMERS + 1):

    customer_id = f"CUST{i:04d}"

    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    customer_name = f"{first_name} {last_name}"

    email = (
        f"{first_name.lower()}."
        f"{last_name.lower()}"
        f"{i}@example.com"
    )

    registration_date = start_registration + timedelta(
        days=random.randint(0, 900)
    )

    customer_type = random.choice(customer_types)

    customers.append({
        "customer_id": customer_id,
        "customer_name": customer_name,
        "email": email,
        "registration_date": registration_date.strftime("%Y-%m-%d"),
        "customer_type": customer_type
    })

customers_df = pd.DataFrame(customers)

print(f"Customers generated: {len(customers_df)}")
display(customers_df.head())
```

### Introduce 2% Invalid Emails

```python
invalid_email_count = int(NUM_CUSTOMERS * 0.02)

invalid_indices = random.sample(
    range(NUM_CUSTOMERS),
    invalid_email_count
)

invalid_emails = [
    "invalidemail.com",
    "missingdomain@",
    "wrong.email",
    "@nodomain.com"
]

for index in invalid_indices:
    customers_df.loc[index, "email"] = random.choice(
        invalid_emails
    )

print(f"Invalid emails introduced: {invalid_email_count}")
```

## 3. Generate Products

```python
products = []

product_counter = 1

for category, subcategories in categories.items():

    for subcategory in subcategories:

        for i in range(NUM_PRODUCTS // 20):

            product_id = f"PROD{product_counter:04d}"

            product_name = (
                f"{subcategory} Product {product_counter}"
            )

            cost_price = round(
                random.uniform(100, 50000),
                2
            )

            products.append({
                "product_id": product_id,
                "product_name": product_name,
                "category": category,
                "subcategory": subcategory,
                "cost_price": cost_price
            })

            product_counter += 1

products_df = pd.DataFrame(products)

print(f"Products generated: {len(products_df)}")
display(products_df.head())
```

### Introduce Product Name Issues

```python
for i in range(len(products_df)):

    if random.random() < 0.10:

        original_name = products_df.loc[i, "product_name"]

        issue_type = random.choice([
            "spaces",
            "lower",
            "upper",
            "mixed"
        ])

        if issue_type == "spaces":
            products_df.loc[i, "product_name"] = (
                "  " + original_name + "  "
            )

        elif issue_type == "lower":
            products_df.loc[i, "product_name"] = (
                original_name.lower()
            )

        elif issue_type == "upper":
            products_df.loc[i, "product_name"] = (
                original_name.upper()
            )

        elif issue_type == "mixed":
            products_df.loc[i, "product_name"] = (
                "  " + original_name.upper() + "  "
            )

print("Product name issues introduced.")
display(products_df.head(10))
```

## 4. Generate Orders

```python
orders = []

start_order_date = datetime(2025, 1, 1)

customer_ids = customers_df["customer_id"].tolist()

for i in range(1, NUM_ORDERS + 1):

    order_id = f"ORD{i:05d}"

    # 5% NULL customer IDs
    if random.random() < 0.05:
        customer_id = None
    else:
        customer_id = random.choice(customer_ids)

    order_date = start_order_date + timedelta(
        days=random.randint(0, 500),
        hours=random.randint(0, 23),
        minutes=random.randint(0, 59),
        seconds=random.randint(0, 59)
    )

    status = random.choice(statuses)
    region_code = random.choice(regions)

    # Some dates intentionally use DD-MM-YYYY format
    if random.random() < 0.05:
        order_date_value = order_date.strftime(
            "%d-%m-%Y %H:%M:%S"
        )
    else:
        order_date_value = order_date.strftime(
            "%Y-%m-%d %H:%M:%S"
        )

    orders.append({
        "order_id": order_id,
        "customer_id": customer_id,
        "order_date": order_date_value,
        "status": status,
        "region_code": region_code
    })

orders_df = pd.DataFrame(orders)

print(f"Orders generated: {len(orders_df)}")
display(orders_df.head())
```

## 5. Generate Order Items

```python
order_items = []

order_ids = orders_df["order_id"].tolist()
product_ids = products_df["product_id"].tolist()

for i in range(1, NUM_ORDER_ITEMS + 1):

    item_id = f"ITEM{i:06d}"

    # Select only existing order IDs
    order_id = random.choice(order_ids)

    product_id = random.choice(product_ids)

    quantity = random.randint(1, 5)

    # 3% negative quantities represent returns
    if random.random() < 0.03:
        quantity = -quantity

    unit_price = round(
        random.uniform(100, 50000),
        2
    )

    discount_percent = round(
        random.uniform(0, 50),
        2
    )

    order_items.append({
        "item_id": item_id,
        "order_id": order_id,
        "product_id": product_id,
        "quantity": quantity,
        "unit_price": unit_price,
        "discount_percent": discount_percent
    })

order_items_df = pd.DataFrame(order_items)

print(f"Order items generated: {len(order_items_df)}")
display(order_items_df.head())
```

## 6. Save the Four CSV Files

```python
customers_df.to_csv(
    "data/customers.csv",
    index=False
)

products_df.to_csv(
    "data/products.csv",
    index=False
)

orders_df.to_csv(
    "data/orders.csv",
    index=False
)

order_items_df.to_csv(
    "data/order_items.csv",
    index=False
)

print("CSV files created successfully:")
print("✓ data/customers.csv")
print("✓ data/products.csv")
print("✓ data/orders.csv")
print("✓ data/order_items.csv")
```
